# Enrollment

```python
import pandas as pd

# Ubah path sesuai lokasi file Anda
file_path = '/content/drive/MyDrive/dataset/skillaju/enrollments.csv'
df1 = pd.read_csv(file_path)
```

```python
df1.head()
```

```python
# Melihat jumlah baris dan kolom
print("Ukuran dataset:", df1.shape)

# Melihat tipe data dan jumlah data non-null
df1.info()
```

```python
# Menghitung missing values setiap kolom
missing_values = df1.isnull().sum()
missing_values
```

```python
# Menghitung jumlah baris duplikat
print("Jumlah duplikat:", df1.duplicated().sum())

print("Jumlah course_id duplikat:", df1['course_id'].duplicated().sum())
print("Jumlah enrollment_id duplikat:", df1['enrollment_id'].duplicated().sum())
print("Jumlah student_id duplikat:", df1['student_id'].duplicated().sum())
```

```python
# Mengubah order_date menjadi datetime
df1['completion_date'] = pd.to_datetime(df1['completion_date'])
df1['enrolled_date'] = pd.to_datetime(df1['enrolled_date'])
df1['last_accessed'] = pd.to_datetime(df1['last_accessed'])
# Memeriksa tipe data
print(df1['completion_date'].dtype)
print(df1['enrolled_date'].dtype)
print(df1['last_accessed'].dtype)
```

```python
df1.head(20)
```

```python
df1['completion_status'] = df1['completion_date'].notna().map({
    True: 'Completed',
    False: 'In Progress'
})
```
